```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# loading cleaned data 

df = pd.read_csv('../Dataset/wine_quality_cleaned.csv')
df.info()
```

```python
# splitting data into X and y 

X = df.drop('quality', axis=1)
y = df['quality']
```

```python
# encoding color column with labelencoder

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
X['color'] = le.fit_transform(X['color'])

X.head(5)
```

```python
# splititng data into training and test set

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, shuffle=True, test_size=0.2, random_state=42)
```

```python
# building a baseline model

from sklearn.ensemble import RandomForestRegressor

baseline_model = RandomForestRegressor(random_state=42)
baseline_model.fit(X_train, y_train)
baseline_model_pred = base_model.predict(X_test)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, root_mean_squared_error

baseline_result = pd.DataFrame({
    'R² Score':[r2_score(y_test, baseline_model_pred)], # How much of the quality score can my model explain?
    'MAE Score': [mean_absolute_error(y_test, baseline_model_pred)], # On average, the prediction is off by about half a quality point
    'RMSE Score': [root_mean_squared_error(y_test, base_model_pred)] # some predictions have larger errors.
},
index=['Baseline Model'])

baseline_result
```

```python
baseline_result.to_csv('../Report/Results/baseline_results.csv', index=False)
```
