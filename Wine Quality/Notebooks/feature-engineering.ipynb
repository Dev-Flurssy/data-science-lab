```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# loading cleaned data 

df = pd.read_csv('../Dataset/wine_quality_cleaned.csv')
df.info()
```

```python
# splitting data into X and y 

X = df.drop('quality', axis=1)
y = df['quality']
```

```python
# encoding color column with labelencoder

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
X['color'] = le.fit_transform(X['color'])

X.head(5)
```

```python
# splititng data into training and test set

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, shuffle=True, test_size=0.2, random_state=42)
```

```python
# creating featured engineering train set.

X_train_fe = X_train.copy()
X_test_fe = X_test.copy()
```

```python
# creating total_acidity column on featured engineering X_train and X_test.

X_train_fe['total_acidity']  = (X_train_fe['fixed_acidity'] + X_train_fe['volatile_acidity'] + X_train_fe['citric_acid'])
X_test_fe['total_acidity'] = (X_test_fe['fixed_acidity'] + X_test_fe['volatile_acidity'] + X_test_fe['citric_acid'])

display(X_train_fe.sample(2))
print()
display(X_test_fe.sample(2))
```

```python
# creating acidity_ratio column on featured engineering X_train and X_test

X_train_fe['acidity_ratio']  = (X_train_fe['fixed_acidity'] / (X_train_fe['volatile_acidity'] + 1e-6)).round(2)
X_test_fe['acidity_ratio'] = (X_test_fe['fixed_acidity'] / (X_test_fe['volatile_acidity']  + 1e-6)).round(2)

display(X_train_fe.sample())
print()
display(X_test_fe.sample())
```

```python
# creating extra_sulphur_dioxide column on featured engineering X_train and X_test.

X_train_fe['extra_sulphur_dioxide']  = (X_train_fe['total_sulfur_dioxide'] - X_train_fe['free_sulfur_dioxide'])
X_test_fe['extra_sulphur_dioxide']  = (X_test_fe['total_sulfur_dioxide'] - X_test_fe['free_sulfur_dioxide'])

display(X_train_fe.sample())
print()
display(X_test_fe.sample())
```

```python
# creating sulphur_dioxide_ratio column on featured engineering X_train and X_test.

X_train_fe['sulphur_dioxide_ratio']  = (X_train_fe['total_sulfur_dioxide'] / (X_train_fe['free_sulfur_dioxide'] + 1e-6)).round(2)
X_test_fe['sulphur_dioxide_ratio']  = (X_test_fe['total_sulfur_dioxide'] / (X_test_fe['free_sulfur_dioxide'] + 1e-6)).round(2)

display(X_train_fe.sample())
print()
display(X_test_fe.sample())
```

```python
# creating sugar_alcohol_ratio column on featured engineering X_train and X_test.

X_train_fe['sugar_alcohol_ratio'] = (X_train_fe['residual_sugar'] / (X_train_fe['alcohol'] + 1e-6)).round(2)
X_test_fe['sugar_alcohol_ratio']  = (X_test_fe['residual_sugar'] / (X_test_fe['alcohol'] + 1e-6)).round(2)

display(X_train_fe.sample())
print()
display(X_test_fe.sample())
```

```python
# creating density_alcohol_ratio column on featured engineering X_train and X_test.

X_train_fe['density_alcohol_ratio'] = (X_train_fe['density'] / (X_train_fe['alcohol'] + 1e-6)).round(2)
X_test_fe['density_alcohol_ratio']  = (X_test_fe['density'] / (X_test_fe['alcohol'] + 1e-6)).round(2)

display(X_train_fe.sample())
print()
display(X_test_fe.sample())
```

```python
# creating alcohol_sulphates column on featured engineering X_train and X_test.

X_train_fe['alcohol_sulphates'] = (X_train_fe['alcohol'] * X_train_fe['sulphates']).round(2)
X_test_fe['alcohol_sulphates']  = (X_test_fe['alcohol']  * X_test_fe['sulphates']).round(2)

display(X_train_fe.sample())
print()
display(X_test_fe.sample())
```

```python
# creating alcohol_acidity column on featured engineering X_train and X_test.

X_train_fe['alcohol_acidity'] = (X_train_fe['alcohol'] * X_train_fe['fixed_acidity']).round(2)
X_test_fe['alcohol_acidity']  = (X_test_fe['alcohol'] * X_test_fe['fixed_acidity']).round(2)

display(X_train_fe.sample())
print()
display(X_test_fe.sample())
```

```python
# saving festured engineer data 

X_train_fe.to_csv('../Dataset/X_train_fe.csv', index=False)
X_test_fe.to_csv('../Dataset/X_test_fe.csv', index=False)
```
