```python
# Loading the Data

import pandas as pd

df = pd.read_csv('../Dataset/wine_quality_raw.csv')
```

```python
# Viewing the shape and info of the data

print(df.shape)
print()
print(df.info())
```

```python
# viewing the dataset

display(df.head())
print()
display(df.tail())
print()
display(df.sample(5))
```

```python
# checking for null values in each column of the dataframe

df.isnull().sum()
```

```python
# checking for duplicates

df.duplicated().sum()
```

```python
# Viewing some samples of duplicates

df[df.duplicated()]
```

```python
# Dropping duplicates 

df.drop_duplicates(inplace=True)
df.info()
```

```python
# Transposed Statisitical Analysis

df.describe().T
```

```python
df['quality'].value_counts()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5), facecolor='darkgrey')


sns.countplot(
    data=df,
    x='quality',
    color='purple'      
)

# Customize title
plt.title(
    'Wine Quality Distribution',
    fontsize=18,
    color='purple',
    fontweight='bold'
)


plt.xlabel('Quality Rating', fontsize=14, color='purple', fontweight='bold')
plt.ylabel('Count', fontsize=14, color='purple', fontweight='bold')


plt.gca().set_facecolor('black') # Change plot area background
plt.show()
```
